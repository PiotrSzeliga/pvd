## Task 3. Data integration

```python
merge1 = rides.merge(users, on='user_id')
```

```python
merge2 = merge1.merge(bikes, on='bike_id')
```

```python
merge3 = merge2.merge(payments, on='ride_id')
```

```python
merge = pd.merge(merge3, stations, left_on='start_station', right_on='station_id').merge(stations, left_on='start_station', right_on='station_id')
```

```python
merge
```

```python
final = merge[['name', 'user_type', 'bike_type', 'station_name_x', 'station_name_y', 'duration_min', 'amount_PLN']]
```

```python
print(pd.concat([final.head(10), final.tail(10)]))
```

## Task 4. User activity analysis

```python
final.groupby('name').aggregate({'user_type':'count','duration_min':'mean'}).rename(columns={'user_type':'ride count', 'duration_min':'average ride duration'}).sort_values(by='ride count', ascending=False).query("`ride count` >= 3")
```
